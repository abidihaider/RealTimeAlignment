```python
from pathlib import Path
import numpy as np

import onnx
from onnx import numpy_helper
```

```python
model = onnx.load("onnx_files_narrow/mlp_int8.onnx")

# Map initializer name -> numpy array
params = {
    init.name: numpy_helper.to_array(init)
    for init in model.graph.initializer
}
params
```

```python
# Extract all initializers (weights, biases, scales, zero-points)
for tensor in model.graph.initializer:
    data = numpy_helper.to_array(tensor) # or int8/uint8 as appropriate
    
    data = data.reshape(tensor.dims)
    print(tensor.name, tensor.dims, tensor.data_type)
    print(data)
```

```python
for node in model.graph.node:
    print(node.op_type, node.input, node.output)
```
